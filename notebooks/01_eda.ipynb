# 01. EDA — LLM Classification

Explore data, check biases, understand distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA_DIR = Path('../data')
train_df = pd.read_csv(DATA_DIR / 'train.csv')
test_df = pd.read_csv(DATA_DIR / 'test.csv')
```

```python
# Shape, columns
print(f'Train: {train_df.shape}, Test: {test_df.shape}')
print(f'\nTrain columns: {train_df.columns.tolist()}')
print(f'\nTest columns: {test_df.columns.tolist()}')
```

```python
# Target distribution
target = (train_df['winner_model_a'].astype(int) * 0 +
          train_df['winner_model_b'].astype(int) * 1 +
          train_df['winner_tie'].astype(int) * 2)
train_df['target'] = target

print('Target distribution:')
print(train_df['target'].value_counts().sort_index())
print('\nProportions:')
print(train_df['target'].value_counts(normalize=True).sort_index())
```

```python
# Length analysis
train_df['len_a'] = train_df['response_a'].str.len()
train_df['len_b'] = train_df['response_b'].str.len()
train_df['len_ratio'] = train_df['len_a'] / (train_df['len_b'] + 1)

print('Length statistics:')
print(train_df[['len_a', 'len_b', 'len_ratio']].describe())

# Length by winner
print('\nAvg length ratio by target:')
print(train_df.groupby('target')['len_ratio'].mean())
```

```python
# Position bias check
# If model_a is listed first, does it win more?
print('Model A wins rate:', train_df['winner_model_a'].mean())
print('Model B wins rate:', train_df['winner_model_b'].mean())
print('Tie rate:', train_df['winner_tie'].mean())
```

```python
# Missing values
print('Train missing:')
print(train_df.isnull().sum())
print('\nTest missing:')
print(test_df.isnull().sum())
```

```python
# Text length distributions
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(train_df['len_a'], bins=50, alpha=0.5, label='response_a')
axes[0].hist(train_df['len_b'], bins=50, alpha=0.5, label='response_b')
axes[0].set_title('Response length distribution')
axes[0].legend()

axes[1].hist(train_df['len_ratio'], bins=50)
axes[1].set_title('Length ratio (A/B)')

plt.tight_layout()
plt.show()
```
